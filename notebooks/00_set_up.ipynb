This script sets the Unity Catalog catalog and schema's needed for the demo to run in the Databricks Free edition

If you are running this in your enterprise environment, you will likely already have these set up, or contact your IT team.

```python
catalog = "dsit_mcp"

# Create a new Unity Catalog
spark.sql(f"CREATE CATALOG IF NOT EXISTS `{catalog}`")
```

```python
# Create a bronze and silver schema in the new UC catalog
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.01_bronze")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.02_silver")
```
